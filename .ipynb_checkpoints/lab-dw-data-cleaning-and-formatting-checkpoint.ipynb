## Exercise 1: Cleaning Column Names

To ensure consistency and ease of use, standardize the column names of the dataframe. Start by taking a first look at the dataframe and identifying any column names that need to be modified. Use appropriate naming conventions and make sure that column names are descriptive and informative.

*Hint*:
- *Column names should be in lower case*
- *White spaces in column names should be replaced by `_`*
- *`st` could be replaced for `state`*

```python
import pandas as pd

# Load insurance company dataset from an online source
url = 'https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv'
df = pd.read_csv(url)
df.head()
```

```python
df.shape
```

```python
df2 = df.copy()
```

```python
#standardize columns name to snake_case
df2.columns = [col.lower().replace(" ","_") for col in df2.columns]
df2.columns
```

```python
#renaming st column 
df2.rename(columns = {'st':"state"}, inplace = True)
df2.columns
```

## Exercise 2: Cleaning invalid Values

The dataset contains columns with inconsistent and incorrect values that could affect the accuracy of our analysis. Therefore, we need to clean these columns to ensure that they only contain valid data.

Note that this exercise will focus only on cleaning inconsistent values and will not involve handling null values (NaN or None).

*Hint*:
- *Gender column contains various inconsistent values such as "F", "M", "Femal", "Male", "female", which need to be standardized, for example, to "M" and "F".*
- *State abbreviations be can replaced with its full name, for example "AZ": "Arizona", "Cali": "California", "WA": "Washington"*
- *In education, "Bachelors" could be replaced by "Bachelor"*
- *In Customer Lifetime Value, delete the `%` character*
- *In vehicle class, "Sports Car", "Luxury SUV" and "Luxury Car" could be replaced by "Luxury"*

```python
df2.gender.unique()
```

```python
gender_mapping = {'Femal': "F", 'female':"F", 'Male': "M"} #we are replacing Femal and female by F and Male by M
df2['gender'] = df2['gender'].map(gender_mapping)
df2.gender.unique()
```

```python
df2.state.unique()
```

```python
state_mapping = {'Washington': "WA", 'Arizona':"AZ", 'California': "CA", "Nevada":"NV", "Oregon":"OR"} #we are replacing very state name by its shortname
df2['state'] = df2['state'].map(state_mapping) #given that we want to replace more than one value it is more efficent to use map
df2.state.unique()
```

```python
df2.education.unique()
```

```python
df2['education'] = df2['education'].str.replace('Bachelors','Bachelor')
df2.education.unique()
```

```python
df2.vehicle_class.unique()
```

```python
df2["vehicle_class"] = df2["vehicle_class"].str.replace("Luxury Car", "Luxury")
df2["vehicle_class"] = df2["vehicle_class"].str.replace("Luxury SUV", "Luxury")
df2["vehicle_class"] = df2["vehicle_class"].str.replace("Sports Car", "Luxury")
```

```python
df2.vehicle_class.unique()
```

```python
df2["customer_lifetime_value"] = df2["customer_lifetime_value"].str.replace("%", "")
df2["customer_lifetime_value"]
```

```python
df2.sample(10)
```

Consider customers with a high policy claim amount to be those in the top 25% of the total claim amount, and clients with a low customer lifetime value to be those in the bottom 25% of the customer lifetime value. Create a pandas DataFrame object that contains information about customers with a policy claim amount greater than the 75th percentile and a customer lifetime value in the bottom 25th percentile.

To do so, select both columns of the dataframe simultaneously and pass it to the `.describe()` method.
```python
#overall description of the dataframe
df2.describe()
```

```python
clv_bottom_thresh = df2["customer_lifetime_value"].quantile(0.25)
clv_bottom_thresh
```

```python
analysis_df.describe()
```
