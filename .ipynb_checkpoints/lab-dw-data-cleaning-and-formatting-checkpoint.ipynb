## Exercise 2: Cleaning invalid Values

The dataset contains columns with inconsistent and incorrect values that could affect the accuracy of our analysis. Therefore, we need to clean these columns to ensure that they only contain valid data.

Note that this exercise will focus only on cleaning inconsistent values and will not involve handling null values (NaN or None).

*Hint*:
- *Gender column contains various inconsistent values such as "F", "M", "Femal", "Male", "female", which need to be standardized, for example, to "M" and "F".*
- *State abbreviations be can replaced with its full name, for example "AZ": "Arizona", "Cali": "California", "WA": "Washington"*
- *In education, "Bachelors" could be replaced by "Bachelor"*
- *In Customer Lifetime Value, delete the `%` character*
- *In vehicle class, "Sports Car", "Luxury SUV" and "Luxury Car" could be replaced by "Luxury"*

```python
# gender column check  values
customer_data['gender'].value_counts()
```

```python
# gender column change
def gender_adjust(gender):
    if gender in ['F', 'female', 'Femal,']:
        return 'F'
    else:
        return 'M'

customer_data['gender'] = customer_data['gender'].apply(gender_adjust)
```

```python
# gender column change check
customer_data['gender'].value_counts()
```

```python
# state column check  values
customer_data['state'].value_counts()
```

```python
# state column change

def state_adjust(state):
    if state == 'AZ':
        return 'Arizona'
    elif state == 'WA':
        return 'Washington'
    elif state == 'Cali':
        return 'California'
    else: 
        return state
    
customer_data['state'] = customer_data['state'].apply(state_adjust)
```

```python
# state column change check
customer_data['state'].value_counts()
```

```python
# education column check values
customer_data['education'].value_counts()
```

```python
# education column change
customer_data['education'] = customer_data['education'].apply(lambda edu: 'Bachelor' if edu == 'Bachelors' else edu)
```

```python
# education column change check
customer_data['education'].value_counts()
```

```python
# customer_lifetime_value column delete % character
customer_data['customer_lifetime_value'] = customer_data['customer_lifetime_value'].str[:-1]
```

```python
# check customer_lifetime_value column delete % character
customer_data['customer_lifetime_value'].value_counts()
```

```python
# vehicle_class column check values
customer_data['vehicle_class'].value_counts()
```

```python
# vehicle_class column change

def car_type_adjust(car):
    if car in ['Sports Car', 'Luxury SUV', 'Luxury Car']:
        return 'Luxury'
    else:
        return car

customer_data['vehicle_class'] = customer_data['vehicle_class'].apply(car_type_adjust)
```

```python
# vehicle_class column change check
customer_data['vehicle_class'].value_counts()
```

## Exercise 3: Formatting data types

The data types of many columns in the dataset appear to be incorrect. This could impact the accuracy of our analysis. To ensure accurate analysis, we need to correct the data types of these columns. Please update the data types of the columns as appropriate.

It is important to note that this exercise does not involve handling null values (NaN or None).

*Hint*:
- *Customer lifetime value should be numeric*
- *Number of open complaints has an incorrect format. Look at the different values it takes with `unique()` and take the middle value. As an example, 1/5/00 should be 5. Number of open complaints is a string - remember you can use `split()` to deal with it and take the number you need. Finally, since it should be numeric, cast the column to be in its proper type.*

```python
# casting customer_lifetime_value to numeric
customer_data['customer_lifetime_value'] = customer_data['customer_lifetime_value'].astype(float)
```

```python
# checking if change was successful
customer_data['customer_lifetime_value'].dtype
```

```python
# number_open_complaints column - checking the colum values
customer_data['number_open_complaints'].unique()
```

```python
# number_open_complaints column - splitting the column
customer_data['number_open_complaints'] = customer_data['number_open_complaints'].str.split('/')
```

```python
# number_open_complaints column - checking the split
customer_data['number_open_complaints'].value_counts()
```

```python
# number_open_complaints column - keeping only the middle value
customer_data['number_open_complaints'] = customer_data['number_open_complaints'].str[1]
```

```python
# number_open_complaints column - checking the keep of middle value
customer_data['number_open_complaints'].value_counts()
```

```python
# number_open_complaints column - setting column as numeric
customer_data['number_open_complaints'] = customer_data['number_open_complaints'].astype(float)
```

```python
# number_open_complaints column - checking the change
customer_data['number_open_complaints'].value_counts()
```
