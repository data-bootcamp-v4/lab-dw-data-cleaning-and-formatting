## Exercise 1: Cleaning Column Names

To ensure consistency and ease of use, standardize the column names of the dataframe. Start by taking a first look at the dataframe and identifying any column names that need to be modified. Use appropriate naming conventions and make sure that column names are descriptive and informative.

*Hint*:
- *Column names should be in lower case*
- *White spaces in column names should be replaced by `_`*
- *`st` could be replaced for `state`*

```python
import pandas as pd
```

```python
df = pd.read_csv("file1.csv")
```

```python
print("First five rows:")
display(df.head())

print("Dataset shape:")
print(df.shape)

print("Original column names:")
print(df.columns.tolist())

print("Dataset information:")
df.info()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

# Rename only the st column
df.rename(columns={"st": "state"}, inplace=True)

print("Cleaned column names:")
print(df.columns.tolist())
```

```python
print("Original gender values:")
print(df["gender"].unique())

gender_mapping = {
    "male": "M",
    "m": "M",
    "female": "F",
    "f": "F",
    "femal": "F"
}

df["gender"] = (
    df["gender"]
    .astype("string")
    .str.strip()
    .str.lower()
    .replace(gender_mapping)
)

print("Cleaned gender values:")
print(df["gender"].value_counts(dropna=False))
```

```python
print("Original state values:")
print(df["state"].unique())

state_mapping = {
    "AZ": "Arizona",
    "WA": "Washington",
    "Cali": "California"
}

df["state"] = df["state"].replace(state_mapping)

print("Cleaned state values:")
print(df["state"].value_counts(dropna=False))
```

```python
print("Original education values:")
print(df["education"].unique())

df["education"] = df["education"].replace({
    "Bachelors": "Bachelor"
})

print("Cleaned education values:")
print(df["education"].value_counts(dropna=False))
```

```python
print("Original customer lifetime values:")
print(df["customer_lifetime_value"].head())

df["customer_lifetime_value"] = (
    df["customer_lifetime_value"]
    .astype("string")
    .str.replace("%", "", regex=False)
    .str.strip()
)

df["customer_lifetime_value"] = pd.to_numeric(
    df["customer_lifetime_value"],
    errors="coerce"
)

print("Cleaned customer lifetime values:")
print(df["customer_lifetime_value"].head())

print("Customer lifetime value data type:")
print(df["customer_lifetime_value"].dtype)
```

```python
print("Original vehicle class values:")
print(df["vehicle_class"].unique())

vehicle_mapping = {
    "Sports Car": "Luxury",
    "Luxury SUV": "Luxury",
    "Luxury Car": "Luxury"
}

df["vehicle_class"] = df["vehicle_class"].replace(
    vehicle_mapping
)

print("Cleaned vehicle class values:")
print(df["vehicle_class"].value_counts(dropna=False))
```

```python
print("Original number of open complaints values:")
print(df["number_of_open_complaints"].unique())

df["number_of_open_complaints"] = (
    df["number_of_open_complaints"]
    .astype("string")
    .str.split("/")
    .str[1]
)

df["number_of_open_complaints"] = pd.to_numeric(
    df["number_of_open_complaints"],
    errors="coerce"
)

print("Cleaned number of open complaints values:")
print(df["number_of_open_complaints"].unique())

print("Number of open complaints data type:")
print(df["number_of_open_complaints"].dtype)
```

```python
print("Null values before handling:")
print(df.isnull().sum())
```

```python
numeric_columns = df.select_dtypes(
    include="number"
).columns

for column in numeric_columns:
    df[column] = df[column].fillna(
        df[column].median()
    )
```

```python
categorical_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for column in categorical_columns:

    mode_value = df[column].mode(dropna=True)

    if not mode_value.empty:
        df[column] = df[column].fillna(
            mode_value.iloc[0]
        )
```

```python
print("Null values after handling:")
print(df.isnull().sum())

print("Total remaining null values:")
print(df.isnull().sum().sum())
```

```python
numeric_columns = df.select_dtypes(
    include="number"
).columns

df[numeric_columns] = df[numeric_columns].astype(int)

print("Updated data types:")
print(df.dtypes)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows before cleaning:")
print(duplicate_count)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Duplicate rows:")
display(duplicate_rows)
```

```python
df.drop_duplicates(
    keep="first",
    inplace=True
)
```

```python
df.reset_index(
    drop=True,
    inplace=True
)
```

```python
print("Number of duplicate rows after cleaning:")
print(df.duplicated().sum())
```

```python
print("Final dataset shape:")
print(df.shape)

print("Final column names:")
print(df.columns.tolist())

print("Final data types:")
print(df.dtypes)

print("Final null value count:")
print(df.isnull().sum())

print("Final duplicate count:")
print(df.duplicated().sum())

print("First five rows of the cleaned dataset:")
display(df.head())
```

```python
df.to_csv(
    "insurance_cleaned.csv",
    index=False
)

print(
    "The cleaned dataset was saved as insurance_cleaned.csv"
)
```

# Bonus: Challenge 2: creating functions on a separate `py` file

Put all the data cleaning and formatting steps into functions, and create a main function that performs all the cleaning and formatting.

Write these functions in separate .py file(s). By putting these steps into functions, we can make the code more modular and easier to maintain.

*Hint: autoreload module is a utility module in Python that allows you to automatically reload modules in the current session when changes are made to the source code. This can be useful in situations where you are actively developing code and want to see the effects of changes you make without having to constantly restart the Python interpreter or Jupyter Notebook kernel.*

```python
%%writefile data_cleaning.py

import pandas as pd


def clean_column_names(df):
    """
    Standardize column names.

    Converts column names to lowercase.
    Removes extra spaces.
    Replaces spaces with underscores.
    Renames st to state.
    """
    df = df.copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
    )

    df = df.rename(columns={
        "st": "state",
        "custateomer_lifetime_value": "customer_lifetime_value"
    })

    return df


def clean_gender(df):
    """
    Standardize gender values to M and F.
    """
    df = df.copy()

    df["gender"] = (
        df["gender"]
        .astype("string")
        .str.strip()
        .str.lower()
        .replace({
            "male": "M",
            "m": "M",
            "female": "F",
            "f": "F",
            "femal": "F"
        })
    )

    return df


def clean_state(df):
    """
    Replace state abbreviations with complete state names.
    """
    df = df.copy()

    state_mapping = {
        "AZ": "Arizona",
        "Cali": "California",
        "WA": "Washington"
    }

    df["state"] = df["state"].replace(state_mapping)

    return df


def clean_education(df):
    """
    Standardize education values.
    """
    df = df.copy()

    df["education"] = df["education"].replace({
        "Bachelors": "Bachelor"
    })

    return df


def clean_customer_lifetime_value(df):
    """
    Remove the percent sign and convert customer lifetime value
    to a numeric data type.
    """
    df = df.copy()

    df["customer_lifetime_value"] = (
        df["customer_lifetime_value"]
        .astype("string")
        .str.replace("%", "", regex=False)
        .str.strip()
    )

    df["customer_lifetime_value"] = pd.to_numeric(
        df["customer_lifetime_value"],
        errors="coerce"
    )

    return df


def clean_vehicle_class(df):
    """
    Combine selected vehicle classes into Luxury.
    """
    df = df.copy()

    df["vehicle_class"] = df["vehicle_class"].replace({
        "Sports Car": "Luxury",
        "Luxury SUV": "Luxury",
        "Luxury Car": "Luxury"
    })

    return df


def clean_open_complaints(df):
    """
    Extract the middle number from values such as 1/5/00.

    Example:
    1/5/00 becomes 5.
    """
    df = df.copy()

    complaints = df["number_of_open_complaints"].astype("string")

    df["number_of_open_complaints"] = pd.to_numeric(
        complaints.str.split("/").str[1],
        errors="coerce"
    )

    return df


def handle_null_values(df):
    """
    Fill numeric null values with the median.

    Fill categorical null values with the mode.
    """
    df = df.copy()

    numeric_columns = df.select_dtypes(include="number").columns

    for column in numeric_columns:
        df[column] = df[column].fillna(df[column].median())

    categorical_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for column in categorical_columns:
        mode_values = df[column].mode(dropna=True)

        if not mode_values.empty:
            df[column] = df[column].fillna(mode_values.iloc[0])

    return df


def convert_numeric_to_integer(df):
    """
    Convert all numeric columns to integer data type.

    This function should be used after handling null values.
    """
    df = df.copy()

    numeric_columns = df.select_dtypes(include="number").columns

    df[numeric_columns] = df[numeric_columns].astype(int)

    return df


def remove_duplicates(df):
    """
    Remove duplicated rows while keeping the first occurrence.

    Reset the index after removing duplicates.
    """
    df = df.copy()

    df = df.drop_duplicates(keep="first")
    df = df.reset_index(drop=True)

    return df


def clean_insurance_data(df):
    """
    Main function that performs all data cleaning steps.
    """
    df = clean_column_names(df)
    df = clean_gender(df)
    df = clean_state(df)
    df = clean_education(df)
    df = clean_customer_lifetime_value(df)
    df = clean_vehicle_class(df)
    df = clean_open_complaints(df)
    df = handle_null_values(df)
    df = convert_numeric_to_integer(df)
    df = remove_duplicates(df)

    return df
```

# Bonus: Challenge 3: Analyzing Clean and Formated Data

You have been tasked with analyzing the data to identify potential areas for improving customer retention and profitability. Your goal is to identify customers with a high policy claim amount and a low customer lifetime value.

In the Pandas Lab, we only looked at high policy claim amounts because we couldn't look into low customer lifetime values. If we had tried to work with that column, we wouldn't have been able to because customer lifetime value wasn't clean and in its proper format. So after cleaning and formatting the data, let's get some more interesting insights!

Instructions:

- Review the statistics again for total claim amount and customer lifetime value to gain an understanding of the data.
- To identify potential areas for improving customer retention and profitability, we want to focus on customers with a high policy claim amount and a low customer lifetime value. Consider customers with a high policy claim amount to be those in the top 25% of the total claim amount, and clients with a low customer lifetime value to be those in the bottom 25% of the customer lifetime value. Create a pandas DataFrame object that contains information about customers with a policy claim amount greater than the 75th percentile and a customer lifetime value in the bottom 25th percentile.
- Use DataFrame methods to calculate summary statistics about the high policy claim amount and low customer lifetime value data. To do so, select both columns of the dataframe simultaneously and pass it to the `.describe()` method. This will give you descriptive statistics, such as mean, median, standard deviation, minimum and maximum values for both columns at the same time, allowing you to compare and analyze their characteristics.

```python
import pandas as pd


# Use the cleaned dataframe if it already exists
data = cleaned_df.copy() if "cleaned_df" in globals() else df.copy()


# Standardize the column names again for safety
data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)


# Correct possible column names damaged by the earlier replacement
data = data.rename(columns={
    "custateomer_lifetime_value": "customer_lifetime_value",
    "total_claim_amount": "total_claim_amount"
})


# Check that the required columns exist
required_columns = [
    "total_claim_amount",
    "customer_lifetime_value"
]

missing_columns = [
    column
    for column in required_columns
    if column not in data.columns
]

if missing_columns:
    print("Missing columns:", missing_columns)
    print("\nAvailable columns:")
    print(data.columns.tolist())

else:
    # Make sure both columns are numeric
    data["total_claim_amount"] = pd.to_numeric(
        data["total_claim_amount"],
        errors="coerce"
    )

    data["customer_lifetime_value"] = pd.to_numeric(
        data["customer_lifetime_value"],
        errors="coerce"
    )


    # Review the summary statistics
    print("Summary statistics for the full dataset:")

    display(
        data[
            [
                "total_claim_amount",
                "customer_lifetime_value"
            ]
        ].describe()
    )


    # Calculate the percentile thresholds
    claim_threshold = data["total_claim_amount"].quantile(0.75)

    clv_threshold = data["customer_lifetime_value"].quantile(0.25)


    print(
        "75th percentile of total claim amount:",
        claim_threshold
    )

    print(
        "25th percentile of customer lifetime value:",
        clv_threshold
    )


    # Filter customers with high claims and low lifetime value
    high_claim_low_clv = data[
        (
            data["total_claim_amount"] > claim_threshold
        )
        &
        (
            data["customer_lifetime_value"] < clv_threshold
        )
    ].copy()


    # Reset the index
    high_claim_low_clv.reset_index(
        drop=True,
        inplace=True
    )


    # Display the filtered dataframe
    print(
        "\nCustomers with high claim amounts and low customer lifetime value:"
    )

    display(high_claim_low_clv)


    # Show the number of customers
    print(
        "Number of selected customers:",
        len(high_claim_low_clv)
    )


    # Summary statistics for the filtered customers
    print(
        "\nSummary statistics for the selected customers:"
    )

    display(
        high_claim_low_clv[
            [
                "total_claim_amount",
                "customer_lifetime_value"
            ]
        ].describe()
    )
```
