```python
import pandas as pd

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)

# Check what the actual column names are
print("Columns before cleaning:")
print(df.columns.tolist())

df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
    .str.replace(r"\bst\b", "state", regex=True)
)

print("Cleaned column names:")
print(df.columns.tolist())

# --- Gender ---
df['gender'] = df['gender'].str.strip().str.lower()
df['gender'] = df['gender'].replace({
    'm': 'M',
    'male': 'M',
    'f': 'F',
    'female': 'F',
    'femal': 'F'
})

# --- State ---
df['state'] = df['state'].replace({
    'AZ': 'Arizona',
    'Cali': 'California',
    'WA': 'Washington'
})

# --- Education ---
df['education'] = df['education'].replace({'Bachelors': 'Bachelor'})

# --- Customer Lifetime Value ---
df['customer_lifetime_value'] = (
    df['customer_lifetime_value']
    .astype(str)
    .str.replace('%', '', regex=False)
    .astype(float)
)

# --- Vehicle Class ---
df['vehicle_class'] = df['vehicle_class'].replace({
    'Sports Car': 'Luxury',
    'Luxury SUV': 'Luxury',
    'Luxury Car': 'Luxury'
})

print("Unique gender values:", df['gender'].unique())
print("Unique state values:", df['state'].unique())
print("Unique education values:", df['education'].unique())
print("Unique vehicle_class values:", df['vehicle_class'].unique())
print(df[['customer_lifetime_value']].head())

# Clean column names (so we can reference them easily)
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

print("Columns:")
print(df.columns.tolist())

print("Customer Lifetime Value unique types:")
print(df['customer_lifetime_value'].dtype)

print("\nSample of 'number_of_open_complaints':")
print(df['number_of_open_complaints'].unique()[:10])

# Convert to numeric
df['customer_lifetime_value'] = (
    df['customer_lifetime_value']
    .astype(str)
    .str.replace('%', '', regex=False)
    .astype(float)
)

# Extract the middle number and convert to numeric
df['number_of_open_complaints'] = (
    df['number_of_open_complaints']
    .astype(str)
    .apply(lambda x: x.split('/')[1] if '/' in x else x)
    .astype(float)
)

print(df[['customer_lifetime_value', 'number_of_open_complaints']].dtypes)
print(df[['customer_lifetime_value', 'number_of_open_complaints']].head())
print(df.dtypes)

null_counts = df.isnull().sum()
print("Missing values per column:")
print(null_counts[null_counts > 0])

# Separate numeric and categorical columns
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

# Fill numeric columns with median
for col in numeric_cols:
    df[col].fillna(df[col].median(), inplace=True)

# Fill categorical columns with mode (most frequent value)
for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("Remaining null values after cleaning:")
print(df.isnull().sum().sum())  # Should print 0

for col in numeric_cols:
    # Convert to integer safely
    df[col] = df[col].astype(int)

print(df.dtypes)

print(df.columns.tolist())

# Identify duplicate rows
duplicates = df.duplicated()

# Count duplicates
print("Number of duplicate rows:", duplicates.sum())

# Optional: display examples of duplicates
print(df[df.duplicated()].head())

# Drop duplicates (keep first occurrence)
df = df.drop_duplicates(keep='first')

# Reset index for consistency
df = df.reset_index(drop=True)

print("Number of duplicate rows after cleaning:", df.duplicated().sum())

df.to_csv("cleaned_no_duplicates-cecile.csv", index=False)
print("✅ Cleaned dataset saved as cleaned_no_duplicates.csv")










```
