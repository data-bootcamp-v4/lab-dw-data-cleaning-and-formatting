```python
def cleaning_columns():
    
    gender_cleaning = {
        'F': 'Female',
        'M': 'Male',
        'Femal': 'Female',
        'Male': 'Male',
        'female': 'Female'
    }

    state_cleaning = {
        'WA': 'Washington',
        'AZ': 'Arizona',
        'Cali': 'California'
    }

    education_cleaning = {
        'Bachelors': 'Bachelor',
        'High School or Below': 'High School'
    }

    df['Gender'] = df['Gender'].map(gender_cleaning)
    df['State'] = df['State'].map(state_cleaning)
    df['Education'] = df['Education'].map(education_cleaning)

    return df
```
