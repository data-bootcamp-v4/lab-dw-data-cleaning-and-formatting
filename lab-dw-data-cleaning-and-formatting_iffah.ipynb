## Exercise 1: Cleaning Column Names

To ensure consistency and ease of use, standardize the column names of the dataframe. Start by taking a first look at the dataframe and identifying any column names that need to be modified. Use appropriate naming conventions and make sure that column names are descriptive and informative.

*Hint*:
- *Column names should be in lower case*
- *White spaces in column names should be replaced by `_`*

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)
```

```python
print("Original column names with quotes:")
for col in df.columns:
    print(f"'{col}'") 
```

```python
print("updated columns name:")
print(df.columns)
```

```python
df.rename(columns={
    'custateomer': 'customer',   
    'custateomer_lifetime_value': 'customer_lifetime_value'
}, inplace=True)
```

```python
print("updated columns name:")
print(df.columns)
```

## Exercise 2: Cleaning invalid Values

The dataset contains columns with inconsistent and incorrect values that could affect the accuracy of our analysis. Therefore, we need to clean these columns to ensure that they only contain valid data.

Note that this exercise will focus only on cleaning inconsistent values and will not involve handling null values (NaN or None).

*Hint*:
- *Gender column contains various inconsistent values such as "F", "M", "Femal", "Male", "female", which need to be standardized, for example, to "M" and "F".*
- *State abbreviations be can replaced with its full name, for example "AZ": "Arizona", "Cali": "California", "WA": "Washington"*
- *In education, "Bachelors" could be replaced by "Bachelor"*
- *In Customer Lifetime Value, delete the `%` character*
- *In vehicle class, "Sports Car", "Luxury SUV" and "Luxury Car" could be replaced by "Luxury"*

```python
df.columns = df.columns.str.lower().str.strip().str.replace(' ', '_')
```

```python
df['gender'] = df['gender'].str.lower().str.strip()
df['gender'] = df['gender'].replace({
    'male': 'M',
    'm': 'M',
    'female': 'F',
    'f': 'F',
    'femal': 'F'
})
```

```python
df['state'] = df['state'].str.strip()
df['state'] = df['state'].replace({
    'AZ': 'Arizona',
    'Cali': 'California',
    'WA': 'Washington'
})
```

```python
df['education'] = df['education'].str.strip()
df['education'] = df['education'].replace({
    'bachelors': 'Bachelor'
})
```

```python
df['customer_lifetime_value'] = df['customer_lifetime_value'].str.replace('%','')
```

```python
df['vehicle_class'] = df['vehicle_class'].str.strip()
df['vehicle_class'] = df['vehicle_class'].replace({
    'Sports Car': 'Luxury',
    'Luxury SUV': 'Luxury',
    'Luxury Car': 'Luxury'
})
```

```python
print("Gender values:", df['gender'].unique())
print("State values:", df['state'].unique())
print("Education values:", df['education'].unique())
print("Customer Lifetime Value sample:", df['customer_lifetime_value'].head())
print("Vehicle Class values:", df['vehicle_class'].unique())
```

## Exercise 4: Dealing with Null values

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)
```

```python
print(df.isnull().sum())
```

```python
numeric_cols = df.select_dtypes(include='number').columns
for col in numeric_cols:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)

categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    mode_value = df[col].mode()[0]  
    df[col] = df[col].fillna(mode_value)
```

```python
print(df.isnull().sum())
```

```python
for col in numeric_cols:
    df[col] = df[col].astype(int)
print(f"{col} converted to int:\n", df[col].head(), "\n")
```
