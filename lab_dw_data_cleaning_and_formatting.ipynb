```python
import pandas as pd
import numpy as np
file1_df = pd.read_csv("../../file_1.csv")
file1_df.head()
```

```python
file1_df.columns = file1_df.columns.str.lower().str.replace(' ', '_')
file1_df = file1_df.rename(columns={'st': 'state'})
file1_df.head()
```

```python
file1_df['number_of_open_complaints'] = file1_df['number_of_open_complaints'].astype(str).str.split('/').str[0]
file1_df.head()
```

```python
gender_map = {
    'female': 'F', 'Femal': 'F', 'f': 'F', 'F': 'F',
    'male': 'M', 'm': 'M', 'Male': 'M', 'M': 'M'
}
file1_df['gender'] = file1_df['gender'].map(gender_map).fillna(file1_df['gender'])

state_map = {
    "AZ": "Arizona",
    "Cali": "California",
    "WA": "Washington"
}
file1_df["state"] = file1_df["state"].replace(state_map)
```

```python
file1_df["education"] = file1_df["education"].replace("Bachelors", "Bachelor")
file1_df["customer_lifetime_value"] = file1_df["customer_lifetime_value"].astype(str).str.replace("%", "")
file1_df["customer_lifetime_value"] = pd.to_numeric(file1_df["customer_lifetime_value"], errors="coerce")

luxury_vehicles = ["Sports Car", "Luxury SUV", "Luxury Car"]
file1_df["vehicle_class"] = file1_df["vehicle_class"].replace(luxury_vehicles, "Luxury")
print("Unique Genders:", file1_df['gender'].unique()) 
print("Unique State:", file1_df["state"].unique())
print("Unique Vehicle Classes:", file1_df["vehicle_class"].unique())
```

```python
file1_df["customer_lifetime_value"] = pd.to_numeric(file1_df["customer_lifetime_value"], errors="coerce")

file1_df["number_of_open_complaints"] = file1_df["number_of_open_complaints"].astype(str).str.split("/").str[1]

file1_df["number_of_open_complaints"] = pd.to_numeric(file1_df["number_of_open_complaints"], errors="coerce")

print(file1_df[["customer_lifetime_value","number_of_open_complaints"]].dtypes)
file1_df[["customer_lifetime_value", "number_of_open_complaints"]].head()
```

```python
print("--- Null Counts Before Cleaning ---")
print(file1_df.isnull().sum())
print("-" * 40)
```

```python
file1_df["gender"] = file1_df["gender"].fillna("Unknown")
file1_df["state"] = file1_df["state"].fillna("Unknown")

clv_mean = file1_df["customer_lifetime_value"].mean()
file1_df["customer_lifetime_value"] = file1_df["customer_lifetime_value"].fillna(clv_mean)

file1_df["number_of_open_complaints"] = file1_df["number_of_open_complaints"].fillna(0)

file1_df["income"] = file1_df["income"].fillna(0)
file1_df["monthly_premium_auto"] = file1_df["monthly_premium_auto"].fillna(0)
```

```python
file1_df["income"] = file1_df["income"].astype(int)
file1_df["monthly_premium_auto"] = file1_df["monthly_premium_auto"].astype(int)
file1_df["number_of_open_complaints"] = file1_df["number_of_open_complaints"].astype(int)
file1_df["customer_lifetime_value"] = file1_df["customer_lifetime_value"].astype(int)

print("--- Null Counts After Cleaning ---")
print(file1_df.isnull().sum())
print("-" * 35)

file1_df.head()
```

```python
duplicate_rows = file1_df[file1_df.duplicated()]
print(f"Number of duplicate rows found: {len(duplicate_rows)}")
```

```python
file1_df = file1_df.drop_duplicates()

file1_df = file1_df.reset_index(drop=True)

print(f"Number of duplicate rows remaining: {file1_df.duplicated().sum()}")

file1_df.to_csv("cleaned_customer_analysis.csv", index=False)
print("Successfully saved cleaned dataset to 'cleaned_customer_analysis.csv'!")
```

```python
# my reasoning for choosing a the particular strategy (drop all duplicate rows completely while keeping their first occurence is based on standard data cleaning best practice) for handling duplicates are,
# duplicate records in customer data typically occur due to system errors, repetitive logging. And in this dataset, the duplicate rows are identical across every single feature(column).
```
