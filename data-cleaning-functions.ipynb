```python
import pandas as pd 
```

```python
# import data

url = 'https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv'
original_df = pd.read_csv(url)
df = original_df.copy()
```

```python
# Cleaning Column Names

def clean_col_names(df):
    df.columns = [ col.lower().replace(" ","_") for col in df.columns ]
    df.rename(columns = {'st':'state'},inplace = True)
    return df

df = clean_col_names(df)
```

```python
# Cleaning invalid Values

def clean_invalid_values(df):
    df['gender_clean'] = df['gender'].map({"Femal":"F","Male":"M","female":"F"})
    df['state_clean'] = df['state'].map({"AZ":"Arizona","Cali":"California","WA": "Washington"})
    df['education_clean'] = df['education'].map({"Bachelors":"Bachelor"})
    df['customer_lifetime_value_clean'] = df['customer_lifetime_value'].apply(lambda x: x if pd.isna(x) else x.replace('%',''))
    return df

df = clean_invalid_values(df)
df['customer_lifetime_value_clean'].unique()
```

```python
# Formatting data types

def format_data_types(df):
    df['customer_lifetime_value_clean'] = df['customer_lifetime_value_clean'].astype("float")
    df['number_of_open_complaints_clean'] = df['number_of_open_complaints'].str.split('/').str[1].astype('Int64')
    return df

df = format_data_types(df)
```

```python
new_df = df[['customer', 'state', 'gender_clean', 'education',
    'customer_lifetime_value_clean', 'income', 'monthly_premium_auto',
    'number_of_open_complaints_clean', 'policy_type',
    'vehicle_class', 'total_claim_amount']]
```

```python
# Handling missing values

def handle_missing_values(series):
    if series.dtypes == 'object':
        return series.fillna('Unknown')
    else:
        return series.fillna(9999)

new_df = new_df.apply(handle_missing_values)
```

```python
# Handling duplicates

def handle_duplicates(df):
    new_df = df.drop_duplicates(keep = False)
    new_df.reset_index(drop=True, inplace=True)
    return new_df

new_df = handle_duplicates(new_df)
new_df.duplicated().sum()
```

```python
# Create a pandas DataFrame object that contains information about customers with a policy claim amount greater than the 75th percentile and a customer lifetime value in the bottom 25th percentile.
condition_percentile_75 = new_df['total_claim_amount'].quantile(0.75)
condition_percentile_25 = new_df['customer_lifetime_value_clean'].quantile(0.25)
df1 = new_df[(new_df['total_claim_amount'] > condition_percentile_75)&(new_df['customer_lifetime_value_clean'] < condition_percentile_25)]
df1
```
