```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)
df
```

```python
df.columns
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df.columns = ['state' if col == 'st' else col for col in df.columns]
```

```python
df.columns
```

```python
def clean_gender(g):
    g = str(g).strip().lower()
    if g in ['f', 'female', 'femal']:
        return 'F'
    elif g in ['m', 'male']:
        return 'M'
    else:
        return g  

df['gender'] = df['gender'].apply(clean_gender)
```

```python
df.head()
```

```python
print(df['state'].unique())
```

```python
state_map = {
    'AZ': 'Arizona',
    'Cali': 'California',
    'WA': 'Washington',
    'CA': 'California' 
}

df['state'] = df['state'].replace(state_map)
```

```python
df.head()
```

```python
def fix_education(ed):
    if isinstance(ed, str) and 'Bachelors' in ed:
        return ed.replace('Bachelors', 'Bachelor')
    return ed

df['education'] = df['education'].apply(fix_education)
```

```python
df['education'].unique()
```

```python
df['customer_lifetime_value'] = df['customer_lifetime_value'].astype(str).str.replace('%', '')
df.head()
```

```python
luxury_map = {
    'Sports Car': 'Luxury',
    'Luxury SUV': 'Luxury',
    'Luxury Car': 'Luxury'
}

df['vehicle_class'] = df['vehicle_class'].replace(luxury_map)
```

```python
df['vehicle_class'].unique()
```

```python
df['customer_lifetime_value'] = pd.to_numeric(df['customer_lifetime_value'], errors='coerce')
```

```python
df['customer_lifetime_value'].dtype
df['customer_lifetime_value'].head()
```

```python
df['number_of_open_complaints'].unique()
```

```python
def extract_middle_number(s):
    if isinstance(s, str):
        parts = s.split('/')
        if len(parts) == 3:
            return parts[1]  
    return s  

df['number_of_open_complaints'] = df['number_of_open_complaints'].apply(extract_middle_number)

df['number_of_open_complaints'] = pd.to_numeric(df['number_of_open_complaints'], errors='coerce')
```

```python
df['number_of_open_complaints'].unique()
```

```python
null_counts = df.isnull().sum()
null_counts
```

```python
# Separating the columns by type
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# for filling numberic columns
for col in num_cols:
    median_val = df[col].median()
    df[col].fillna(median_val, inplace=True)

# for filling categorical columns
for col in cat_cols:
    mode_val = df[col].mode()[0]
    df[col].fillna(mode_val, inplace=True)
```

```python
df.isnull().sum()
```

```python
for col in num_cols:
    df[col] = df[col].astype(int)
```

```python
df.dtypes
```

```python
duplicates = df.duplicated()
duplicates.sum()
```

```python
df_clean = df.drop_duplicates(keep='first').reset_index(drop=True)
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.to_csv('cleaned_data.csv', index=False)
```
